#### Interpolating

```python
csv_1 = pd.read_csv("C:\\Users\\krush\\Downloads\\HRDataset_v14.csv")
csv_1
```

```python
csv_1.interpolate()    # it works only on numbers it dosnt work on alphabets
```

```python
# following are the methods of interpolate function
csv_2 = csv_1.interpolate(method="linear")
print(csv_2)
```

```python
csv_2 = csv_1.interpolate(method="linear",axis=0)  # axis 0=y-axis, axis 1 = x-axis
print(csv_2)
```

```python
# limit method- limiting the number of updates
csv_2 = csv_1.interpolate(limit=2,limit_direction="forward",axis=0)
print(csv_2)
```

```python
# limit area

csv_3 = csv_1.interpolate(limit_area="inside")
print(csv_3)
# inside - fill all the null data inside the dataframe
# outside - dosnt fill any null data/ no filling occures
```

```python
# inplace method
# If inplae = True --> there is change in original dataset
# inplace = False --> there is a copy of original dataset will produce and it will update accordingly

csv_4 = csv_1.interpolate(limit_direction="both",limit=2,inplace=False)
print(csv_4)
```

### Merger of 2 data_frames and csv files

```python
var1 = pd.DataFrame({"a":[1,3,2,5],"b":[6,5,9,6]})
var2 = pd.DataFrame({"a":[1,3,2,5],"c":[9,4,3,7]})

print(pd.merge(var1, var2, on="a"))

#there will be the merger of similar datatypes
var1 = pd.DataFrame({"a":[1,3,2,5],"b":[6,5,9,6]})
var2 = pd.DataFrame({"a":[1,3,2,0],"c":[9,4,3,7]})

print(pd.merge(var1, var2, on="a"))
```

```python
# merger with missing data
var3 = pd.merge(var1, var2, how = "left")
print(var3)
var4 = pd.merge(var1, var2, how = "right")
print(var4)
var5 = pd.merge(var1, var2, how = "outer", indicator=True)  # indicator is for finding which data is missing 
print(var5)
```

```python
var1 = pd.DataFrame({"A":[1,3,2,5],"B":[6,5,9,6]})
var2 = pd.DataFrame({"A":[1,3,2,5],"B":[9,4,3,7]})

pd.merge(var1, var2, left_index=True, right_index=True, suffixes=("name", "ID"))
```

```python
# Concat
# series
import pandas as pd
s1 = pd.Series([1,2,3,4,5,6])
s2 = pd.Series([17,25,34,43,52,61])

pd.concat([s1,s2])
```
