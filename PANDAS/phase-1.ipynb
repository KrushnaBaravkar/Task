## Data Frames

```python
# creating the dataframe using list and dict

l = [[1,3,2,4,5],[1,7,5,6,9]]

data_frame = pd.DataFrame(l, index=[1,3])
print("Data frame by List:\n",data_frame)

d = {'a':[1,3,2],
     'b':[4,0,7],
     'c':[5,6,0]}

data_frame1 = pd.DataFrame(d)
print("data frame by dictionare:\n", data_frame1)
print("required coloumn:\n", data_frame1["a"])
print("indexing the rows:\n", pd.DataFrame(d, index=['a','b','c']))
```

```python
# adding the column

data_frame1['d'] = [0,7,5]
print(data_frame1)

# modifing the columns
data_frame1['a'] = data_frame1['a'] + 1
print("modified column a:\n", data_frame1)

# deleting and renaming the columns
data_frame1.drop(columns=['d'], inplace=True) # dropping column
data_frame1.rename(columns={'a':'new'}, inplace=True)  # renaming
print("\n \n", data_frame1)

# getting perticular data
print("\n element:", data_frame1.iat[1,1])
```

```python
# data frames from series
sr = {"s":pd.Series([1,2,3,4]),"r":pd.Series([2,5,3,6])}

var = pd.DataFrame(sr)

print(type(var))
print(var)
```

### Arthematic operatins


```python
var = pd.DataFrame({"a":[1,2,3,4,5],"b":[5,6,7,8,9]})
var

# addition of column a and column b
var["c = a+b"] = var["a"] + var["b"]
print("column with operation:\n", var)

var["d = a-b"] = var["a"] - var["b"]
print("column with operation:\n", var)

# also we can divid and multiply thode columns
```

```python
var = pd.DataFrame({"a":[10,20,30,40,50],"b":[15,16,17,18,19]})
var

var["c"] = var["a"] >= 25
var["d"] = var["b"] <= 17
print(var)

```
